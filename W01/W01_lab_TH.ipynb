## 2. นิพจน์และตัวดำเนินการ

| ตัวดำเนินการ | ความหมาย | ตัวอย่าง | ผลลัพธ์ |
|---|---|---|---|
| `+` `-` `*` | บวก ลบ คูณ | `55 * 2` | `110` |
| `/` | หาร ผลลัพธ์เป็น float เสมอ | `10 / 2` | `5.0` |
| `//` | หารปัดเศษลง (floor division) | `7 // 2` | `3` |
| `%` | เศษจากการหาร (modulo) ไม่ใช่ร้อยละ | `7 % 2` | `1` |
| `**` | ยกกำลัง | `2 ** 10` | `1024` |

ลำดับการคำนวณ: `**` ก่อน, ตามด้วย `*` `/` `//` `%`, แล้วจึง `+` `-` ตัวดำเนินการระดับเดียวกันคำนวณจากซ้ายไปขวา ใช้วงเล็บเมื่อต้องการกำหนดลำดับเอง หรือเพื่อให้อ่านง่าย

````
print(2 + 3 * 4)
print((2 + 3) * 4)
print(100 * 2 * (1 - 0.10))   # ราคาหลังลด 10% ของสินค้า 2 ชิ้น ชิ้นละ 100 บาท
````

`//` และ `%` ใช้กับการแบ่งเป็นจำนวนเต็ม เช่น งบ 200 บาท ซื้อสินค้าราคา 55 บาท ได้ 3 ชิ้น เหลือ 35 บาท เพราะ 200 = 3 × 55 + 35

````
budget = 200
price = 55
print(budget // price, budget % price)
````

## 6. การจัดรูปแบบผลลัพธ์ด้วย f-string

f-string แทนค่าตัวแปรในข้อความผ่าน `{ }` และกำหนดรูปแบบหลังเครื่องหมาย `:`

| รูปแบบ | ความหมาย |
|---|---|
| `{x:.2f}` | ทศนิยม 2 ตำแหน่ง |
| `{x:,.2f}` | คั่นหลักพันและทศนิยม 2 ตำแหน่ง |
| `{s:<12}` / `{x:>12}` | ชิดซ้าย / ชิดขวา ในความกว้าง 12 ตัวอักษร |

การจัดความกว้างนับตามจำนวนอักขระ สระบนล่างและวรรณยุกต์ภาษาไทยจึงทำให้คอลัมน์เหลื่อม ตารางที่ต้องจัดแนวจึงควรใช้ข้อความภาษาอังกฤษ

````
total = 1234567.891
print(f"ยอดขายรวม {total:,.2f} บาท")
print(f"[{'Latte':<12}]")
print(f"[{110:>12,.2f}]")
````

## 7. ตัวอย่าง: ใบเสร็จ

ลาเต้ 2 แก้ว แก้วละ 55 บาท อเมริกาโน่ 2 แก้ว แก้วละ 45 บาท ส่วนลด 10% แล้วคิด VAT 7% จากยอดหลังส่วนลด

อัตราทั้งหมดเก็บเป็นตัวแปรไว้ที่ต้นโค้ด เมื่ออัตราเปลี่ยน แก้ค่าเพียงจุดเดียวแล้วรันใหม่ ทุกบรรทัดจะคำนวณตาม นี่คือข้อได้เปรียบหลักของการเขียนโค้ดเมื่อเทียบกับการคำนวณทีละครั้งด้วยเครื่องคิดเลข

````
latte_price = 55
americano_price = 45
latte_qty = 2
americano_qty = 2
discount_rate = 0.10
vat_rate = 0.07

latte_total = latte_price * latte_qty
americano_total = americano_price * americano_qty
subtotal = latte_total + americano_total
discount = subtotal * discount_rate
after_discount = subtotal - discount
vat = after_discount * vat_rate
grand_total = after_discount + vat

print("=" * 30)
print("     Campus Coffee Receipt")
print("=" * 30)
print(f"{'Latte':<12}x{latte_qty:<3}{latte_total:>14,.2f}")
print(f"{'Americano':<12}x{americano_qty:<3}{americano_total:>14,.2f}")
print("-" * 30)
print(f"{'Subtotal':<16}{subtotal:>14,.2f}")
print(f"{'Discount 10%':<16}{-discount:>14,.2f}")
print(f"{'After discount':<16}{after_discount:>14,.2f}")
print(f"{'VAT 7%':<16}{vat:>14,.2f}")
print("-" * 30)
print(f"{'TOTAL':<16}{grand_total:>14,.2f}")
````
