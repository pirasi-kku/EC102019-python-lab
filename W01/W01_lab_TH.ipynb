````python
#@title ตัวช่วยตรวจคำตอบ (กดรันเซลล์นี้ก่อน ไม่ต้องแก้) { display-mode: "form" }
# คำตอบถูกเก็บแบบเข้ารหัส (hash) โค้ดนี้ตรวจได้ว่าถูกหรือไม่ แต่ไม่เก็บตัวเลขเฉลยไว้
import hashlib as _hl
_KEY = {
    'ข้อ 1': '84a4488cd1d2b4c167fa',
    'ข้อ 2 แก้ว': '5b924369e39681cc86cb',
    'ข้อ 2 เงินเหลือ': 'e99e3fe741d8f88eac13',
    'ข้อ 3 ยอดขาย': '451fcd80c5fa1e17cc4d',
    'ข้อ 3 เงินเฟ้อ': 'e699314a73b86e0b9f27',
    'ข้อ 4 ก่อนภาษี': '8fa1bfd8d31d703e0855',
    'ข้อ 4 VAT': '7d0023f1ffe60a7dbbbe',
    'ข้อ 5 ยอดสุทธิ': '7cd0867c42171ddd0031',
    'ข้อ 5 ข้อความ': 'a4712de37258d4aa8c80',
    'ท้าทาย': 'e55b2b4b6b75ed63ac60'
}
_RESULTS = {}

def _canon(x):
    if hasattr(x, "tolist") and not isinstance(x, str):
        x = x.tolist()
    if isinstance(x, bool):
        return "b:" + str(x)
    if isinstance(x, (int, float)):
        v = round(float(x), 2) + 0.0
        if abs(v) < 0.005:
            v = 0.0
        return "n:%.2f" % v
    if isinstance(x, str):
        return "s:" + x
    if isinstance(x, (list, tuple)):
        return "l:" + "|".join(_canon(e) for e in x)
    return "o:" + repr(x)

def _h(label, c):
    return _hl.sha256(f"EC102019-2569|{label}|{c}".encode()).hexdigest()[:20]

def check(label, answer, tol=0.01):
    target = _KEY.get(label)
    if target is None:
        print(f"[{label}] ไม่พบข้อนี้ในเฉลย")
        return
    ok = False
    if answer is ...:
        print(f"[{label}] ยังไม่ได้เติมคำตอบ (ยังเป็น ...)")
        _RESULTS[label] = False
        return
    a = answer.tolist() if hasattr(answer, "tolist") and not isinstance(answer, str) else answer
    if isinstance(a, (int, float)) and not isinstance(a, bool):
        base = round(float(a), 2)
        k = int(tol / 0.01) + 1
        for i in range(-k, k + 1):
            c = round(base + i * 0.01, 2)
            if abs(a - c) <= tol + 0.005 and _h(label, _canon(c)) == target:
                ok = True
                break
    else:
        ok = _h(label, _canon(a)) == target
    if ok:
        print(f"[{label}] ถูกต้อง")
    elif isinstance(a, str):
        print(f"[{label}] " + 'ยังไม่ถูก ได้ {a!r} ตรวจตัวสะกด ช่องว่าง และตัวพิมพ์เล็ก-ใหญ่'.format(a=a))
    else:
        print(f"[{label}] " + 'ยังไม่ถูก ได้ {a} ลองตรวจสูตรอีกครั้ง'.format(a=answer))
    _RESULTS[label] = bool(ok)

def _group(k):
    if k.startswith(("ท้าทาย", "Challenge")):
        return 2
    if k.startswith(("งานจริง", "Real")):
        return 1
    return 0

def summary():
    heads = ['สรุปแบบฝึกที่ผ่าน (ข้อ 1–5 ใช้ให้คะแนน): {p} จาก {n} ข้อย่อย', 'งานจริงที่ผ่าน: {p} จาก {n} ข้อย่อย', 'โจทย์ท้าทาย (ไม่มีคะแนน): {p} จาก {n} ข้อย่อย']
    for g in range(3):
        keys = [k for k in _KEY if _group(k) == g]
        if not keys:
            continue
        passed = sum(1 for k in keys if _RESULTS.get(k))
        print(heads[g].format(p=passed, n=len(keys)))
        for k in keys:
            print("  " + ('ผ่าน' if _RESULTS.get(k) else 'ยังไม่ผ่าน') + "  " + k)

print('พร้อมตรวจคำตอบแล้ว')
````
