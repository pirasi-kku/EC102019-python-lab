## 1. Percentages and the base of comparison

$$\text{Percentage change} = \frac{\text{new} - \text{old}}{\text{old}} \times 100$$

A percentage is meaningful only relative to its base. An item priced at 100 baht, discounted by 10% and then marked up by 10% of the discounted price, ends at 99 baht, not 100, because the second change is computed on a base of 90.

Likewise, a rate of 7% is 0.07: 120 baht plus 7% VAT is 120 × 1.07, and a VAT-inclusive price must be divided by 1.07 to recover the pre-VAT price.

````
100 * 0.90 * 1.10   # the small excess at the end is explained in Section 4
````

## 2. Expressions and operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | add, subtract, multiply | `55 * 2` | `110` |
| `/` | division, always returns a float | `10 / 2` | `5.0` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | remainder (modulo), not percent | `7 % 2` | `1` |
| `**` | power | `2 ** 10` | `1024` |

Precedence: `**` first, then `*` `/` `//` `%`, then `+` `-`. Operators of equal precedence are evaluated left to right. Use parentheses to set the order explicitly or to make an expression easier to read.

````
print(2 + 3 * 4)
print((2 + 3) * 4)
print(100 * 2 * (1 - 0.10))   # two items at 100 baht each, 10% off
````

`//` and `%` handle whole-unit allocation: with 200 baht and an item at 55 baht you can buy 3 and keep 35, since 200 = 3 × 55 + 35.

````
budget = 200
price = 55
print(budget // price, budget % price)
````

## 3. Variables and assignment

`=` is assignment: Python evaluates the right-hand side first and binds the result to the name on the left. It is not a mathematical equation, so `x = x + 1` means "take the current value of `x`, add 1, and bind the result to `x`". The old value is replaced.

````
latte_price = 55
latte_qty = 2
latte_total = latte_price * latte_qty
latte_total
````

````
x = 5
x = x + 1
x = x * 2
x
````

**Naming:** letters, digits and `_`, not starting with a digit. Names are case-sensitive (`Price` and `price` differ). Python's style guide (PEP 8) uses lowercase words joined by `_`, such as `vat_rate`; choose names that say what the value means.

## 4. Basic data types

- `int`: whole numbers, e.g. quantities
- `float`: decimal numbers, e.g. tax rates, discounted prices
- `str`: text, written in quotes

Operators behave according to type: `"55" * 2` repeats the text, while `"55" + 5` raises a `TypeError` because text and numbers cannot be added.

````
print(type(55), type(55.0), type("55"))
print("55" * 2)
print(55 * 2)
````

````
"55" + 5
````

**Floating-point numbers:** a `float` is stored in binary following IEEE 754. Values such as 0.1 have no exact binary representation, so small errors appear; this is not a calculation mistake. Monetary output is therefore rounded to 2 decimal places. For accounting work that must be exact to the satang at every step, Python provides the `decimal` module.

````
print(0.1 + 0.2)
print(round(0.1 + 0.2, 2))

from decimal import Decimal
print(Decimal("0.1") + Decimal("0.2"))
````

## 5. Writing numbers in code

- Percentages are written as decimals: 7% is `0.07`. Writing `7%` raises a `SyntaxError`.
- Do not use commas as thousands separators: `42,000` raises no error, but Python reads it as two values, `(42, 0)`. A silent wrong result is more dangerous than an error. For readability, write `42_000`.

````
vat_rate = 7%
````

````
x = 42,000
print(x, type(x))

y = 42_000
print(y, type(y))
````

## 6. Formatting output with f-strings

An f-string substitutes values inside `{ }`; the format follows the `:`.

| Format | Meaning |
|---|---|
| `{x:.2f}` | 2 decimal places |
| `{x:,.2f}` | thousands separator and 2 decimal places |
| `{s:<12}` / `{x:>12}` | left / right aligned in a width of 12 characters |

````
total = 1234567.891
print(f"Total sales {total:,.2f} baht")
print(f"[{'Latte':<12}]")
print(f"[{110:>12,.2f}]")
````

## 7. Example: a receipt

Two lattes at 55 baht, two americanos at 45 baht, a 10% discount, then 7% VAT on the discounted amount.

All rates are stored as variables at the top. When a rate changes, edit one value and rerun; every line is recomputed. This is the main advantage of code over one-off calculation with a calculator.

````
latte_price = 55
americano_price = 45
latte_qty = 2
americano_qty = 2
discount_rate = 0.10
vat_rate = 0.07

latte_total = latte_price * latte_qty
americano_total = americano_price * americano_qty
subtotal = latte_total + americano_total
discount = subtotal * discount_rate
after_discount = subtotal - discount
vat = after_discount * vat_rate
grand_total = after_discount + vat

print("=" * 30)
print("     Campus Coffee Receipt")
print("=" * 30)
print(f"{'Latte':<12}x{latte_qty:<3}{latte_total:>14,.2f}")
print(f"{'Americano':<12}x{americano_qty:<3}{americano_total:>14,.2f}")
print("-" * 30)
print(f"{'Subtotal':<16}{subtotal:>14,.2f}")
print(f"{'Discount 10%':<16}{-discount:>14,.2f}")
print(f"{'After discount':<16}{after_discount:>14,.2f}")
print(f"{'VAT 7%':<16}{vat:>14,.2f}")
print("-" * 30)
print(f"{'TOTAL':<16}{grand_total:>14,.2f}")
````
